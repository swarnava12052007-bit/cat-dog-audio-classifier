```python
import logging
import os
from datetime import datetime

# Custom function to bypass the from_root library bug in Jupyter Notebooks
def from_root():
    return os.path.abspath(os.path.join(os.getcwd(), "../../.."))
```

```python
LOG_FILE = f"{datetime.now().strftime('%m_%d_%Y_%H_%M_%S')}.log"
logs_path = os.path.join(from_root(), "logs", LOG_FILE)
```

```python
os.makedirs(logs_path, exist_ok=True)
LOG_FILE_PATH = os.path.join(logs_path, LOG_FILE)
```

```python
logging.basicConfig(filename=LOG_FILE_PATH, format="[%(asctime)s] %(name)s - %(levelname)s - %(message)s",
                    level=logging.DEBUG)
```
